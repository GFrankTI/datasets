```python
import pandas as pd

datos = [
    [1, "Leche", 2],
    [1, "Pan", 1],
    [1, "Queso", 1],
    [2, "Leche", 1],
    [2, "Pan", 2],
    [3, "Pan", 1],
    [3, "Queso", 1],
    [4, "Leche", 1],
    [4, "Pan", 1],
    [5, "Leche", 1],
    [5, "Queso", 2],
    [6, "Pan", 1],
    [6, "Queso", 1],
    [7, "Leche", 1],
    [7, "Pan", 1],
    [7, "Queso", 1],
    [8, "Huevos", 2],
    [8, "Pan", 1],
    [9, "Leche", 1],
    [9, "Pan", 1],
    [10, "Huevos", 1],
    [10, "Queso", 1],
]

df_ventas = pd.DataFrame(
    datos,
    columns=["venta_id", "producto", "cantidad"]
)

print(df_ventas)
```

```python
transacciones = (
    df_ventas.groupby("venta_id")["producto"]
    .agg(list)
)

print(transacciones)
```

```python
df_binario = pd.crosstab(
    df_ventas["venta_id"],
    df_ventas["producto"]
)#.gt(0)

print(df_binario)
```

```python
soportes = df_binario.mean()

print(soportes)
```

```python
soporte_minimo = 0.30

frecuentes_1 = soportes[
    soportes >= soporte_minimo
]

print(frecuentes_1)
```

```python
from itertools import combinations

productos = frecuentes_1.index

for a, b in combinations(productos, 2):
    soporte = (df_binario[a] & df_binario[b]).mean()
    print(a, b, soporte)
```

```python
from mlxtend.frequent_patterns import apriori

frecuentes = apriori(
    df_binario,
    min_support=0.30,
    use_colnames=True
)

print(frecuentes)
```

```python
from mlxtend.frequent_patterns import association_rules

reglas = association_rules(
    frecuentes,
    metric="confidence",
    min_threshold=0.50
)

print(
    reglas[[
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]]
)
```
